Q1: NLP Preprocessing Pipeline

```python
# Import required libraries
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer


def preprocess_text(sentence):
    # Step 1: Tokenization
    tokens = word_tokenize(sentence)
    print("Original Tokens:", tokens)

    # Step 2: Remove stopwords and punctuation
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [
        word for word in tokens
        if word.lower() not in stop_words and word.isalpha()
    ]
    print("Tokens Without Stopwords:", filtered_tokens)

    # Step 3: Stemming
    stemmer = PorterStemmer()
    stemmed_words = [
        stemmer.stem(word) for word in filtered_tokens
    ]
    print("Stemmed Words:", stemmed_words)


sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."

preprocess_text(sentence)
```

Short Answer Questions

1- What is the difference between stemming and lemmatization? Provide examples with the word “running.”

Stemming removes word endings and may produce invalid words, while lemmatization uses grammar and a dictionary to find the correct base word. For example, both Porter stemming and verb-aware lemmatization convert "running" to "run."

2- Why might removing stop words be useful in some NLP tasks, and when might it actually be harmful?

Removing stopwords can make NLP processing faster by removing unnecessary words. However, it can be harmful when important words like "not" are removed.

Q2: Named Entity Recognition with SpaCy

```python

import spacy

# Load the English language model
nlp = spacy.load("en_core_web_sm")

# Input sentence
sentence = "LeBron James as the NBA player of the United States in Laker and won the NBA Final MVP in 2020."

# Process the sentence
doc = nlp(sentence)

# Print entity text, label, and character positions
for ent in doc.ents:
    print("Entity:", ent.text)
    print("Label:", ent.label_)
    print("Start:", ent.start_char)
    print("End:", ent.end_char)
    print("----------------------")
```

1- How does NER differ from POS tagging in NLP?

NER identifies named entities such as people, locations, andorganizations.
POS tagging identifies the grammatical role of each word, such as nouns, verbs, and adjectives.

Q3: Scaled Dot-Product Attention:

```python

import numpy as np

# Scaled dot-product attention function
def scaled_dot_product_attention(Q, K, V):

    # Step 1: Compute QK^T
    scores = np.dot(Q, K.T)

    # Step 2: Scale by square root of key dimension
    d = K.shape[1]
    scaled_scores = scores / np.sqrt(d)

    # Step 3: Apply softmax
    exp_scores = np.exp(
        scaled_scores - np.max(scaled_scores, axis=-1, keepdims=True)
    )
    attention_weights = exp_scores / np.sum(
        exp_scores, axis=-1, keepdims=True
    )

    # Step 4: Multiply attention weights by V
    output = np.dot(attention_weights, V)

    return attention_weights, output


# Test input matrices
Q = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]])

K = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]])

V = np.array([[1, 2, 3, 4],
              [5, 6, 7, 8]])

# Call the function
weights, output = scaled_dot_product_attention(Q, K, V)

# Display results
print("Attention Weights Matrix:")
print(np.round(weights, 3))

print("\nFinal Output Matrix:")
print(np.round(output, 3))
```

Short Answer Questions:

1- Why do we divide the attention score by √d in the scaled dot-product attention formula?

We divide attention scores by √d to prevent large values and preserve gradients during training.

2- How does self-attention help the model understand relationships between words in a sentence?

Self-attention allows each word to focus on other relevant words in a sentence to better understand their relationships and meaning.
